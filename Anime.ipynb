```python
 #Normalize text data#
df['synopsis'] = df['synopsis'].apply(lambda x: x.lower().replace('\n',' ').split())
```

```python
anime_data = df[['mal_id', 'titles','score','synopsis', 'genres', 'themes' ]]
```

```python
anime_data.head()
```

```python
anime_data['titles'].apply(lambda x: ' '.join(x))
```

```python
anime_data['titles'] = anime_data['titles'].apply(lambda x: x[0])
```

```python
anime_data['score'] = anime_data['score'].apply(lambda x:[str(x)])
```

```python
def genre_space(value):
    temp = []
    for genre in value:
        genre = genre.replace(' ','')
        temp.append(genre)
    return temp
```

```python
genre_space(["Avant Garde", "Comedy"])
```

```python
anime_data['genres'] = anime_data['genres'].apply(genre_space)
```

```python
anime_data['themes'] = anime_data['themes'].apply(genre_space)
```

```python
anime_data['Tag'] = anime_data['score'] + anime_data['genres'] + anime_data['themes'] + anime_data['synopsis']
```

```python
anime_data = anime_data[['mal_id','titles','Tag']]
```

```python
anime_data.to_csv('Anime_data.csv')
```

```python
anime_data.head()
```

```python
anime_data['Tag'] = anime_data['Tag'].apply(lambda x:' '.join(x).lower())
```

```python
from nltk.stem.porter import PorterStemmer

stemming = PorterStemmer()
```

```python
def stemming_words(items):
  temp_lst = []
  items = items.split()
  for word in items:
    temp_lst.append(stemming.stem(word))
  return temp_lst
```

```python
anime_data['Tag'] =anime_data['Tag'].apply(stemming_words)
```

```python
anime_data['Tag'] = anime_data['Tag'].apply(lambda item: ' '.join(item))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF Vectorization for 'synopsis'
tfidf = TfidfVectorizer(stop_words='english', max_features=5000)
tfidf_matrix = tfidf.fit_transform(anime_data['Tag'])
```

```python
tfidf_matrix = tfidf_matrix.toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similar_matrix = cosine_similarity(tfidf_matrix)
```

```python
sorted(similar_matrix[0],reverse=True)[1:5]
```

```python
def recommend(anime):
    try:
        index = anime_data[anime_data['titles'] == anime].index[0]
        distances = sorted(list(enumerate(similar_matrix[index])),
                           reverse=True,key = lambda x: x[1])
        for i in distances[1:6]:
            print(anime_data.iloc[i[0]]['titles'])
    except:
        records = anime_data[anime_data['titles'].str.contains(anime)]['titles']
        print(records)
        index = int(input('Enter Index of the Anime : '))
        distances = sorted(list(enumerate(similar_matrix[index])),
                           reverse=True,key = lambda x: x[1])
        for i in distances[1:6]:
            print(anime_data.iloc[i[0]]['titles'])
```

```python
recommend('One')
```
